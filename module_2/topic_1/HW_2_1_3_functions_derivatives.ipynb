1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
x = np.linspace(0, 10, 100) # 100 значень у діапазоні від 0 до 10
f = 10 * x - 5 # задана функція
df = np.full_like(x, 10) # масив тієї ж форми, що й x, та заповнює його числом 10
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, f, label='f(x) = 10x - 5')
plt.plot(x, df, label="f'(x) = 10", linestyle='--')

plt.xlabel('Кількість годин')
plt.ylabel('Оцінка')
plt.title('Функція')

plt.grid(True)
plt.legend()

plt.show()
```

> - `f(x)` -> пряма лінія з постійним зростанням
> - `f′(x) = 10` -> горизонтальна лінія, що показує сталу швидкість зміни
> - кожна додаткова година навчання завжди підвищує оцінку на однакову величину`

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

# Функції
f1 = sp.sin(x)
f2 = sp.log(x)

# Похідні
df1 = sp.diff(f1, x)
df2 = sp.diff(f2, x)

print(f'f(x) = {f1}, f`(x) = {df1}')
print(f'f(x) = {f2}, f`(x) = {df2}')
```

```python
# Перетворення
f1_func = sp.lambdify(x, f1, 'numpy')
df1_func = sp.lambdify(x, df1, 'numpy')

f2_func = sp.lambdify(x, f2, 'numpy')
df2_func = sp.lambdify(x, df2, 'numpy')
```

```python
# Дані
x1 = np.linspace(0, 2*np.pi, 400) # для sin(x)
x2 = np.linspace(0.1, 10, 1000) # для ln(x), x > 0
```

```python
plt.figure(figsize=(8, 5))

# sin(x) та cos(x)
plt.plot(x1, f1_func(x1), label='sin(x)')
plt.plot(x1, df1_func(x1), label="cos(x)")
plt.title("f(x) = sin(x) та її похідна")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# ln(x) та 1/x
plt.plot(x2, f2_func(x2), label='ln(x)')
plt.plot(x2, df2_func(x2), label='1/x')
plt.title("f(x) = ln(x) та її похідна")

plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x**2 - 5*x + 1

df = sp.diff(f, x)

f_func = sp.lambdify(x, f, 'numpy')
df_func= sp.lambdify(x, df, 'numpy')

x_data = np.linspace(0, 100, 500)

y = f_func(x_data)
y_df = df_func(x_data)

print(f'f(x) = {f}, f`(x) = {df}')
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x_data, y, label='f(x) = x^2 − 5x + 1')
plt.plot(x_data, y_df, label="f`(x) = 2x − 5")

plt.xlabel('X')
plt.ylabel('Значення')
plt.title('Функція та її похідна')

plt.legend()
plt.grid(True)

plt.show()
```
