```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
# Task 1
result_matrix  = A @ B
# matrix_dot_product_numpy = np.dot(A, B) # через numpy
print(result_matrix)
print(f'Форма - {result_matrix.shape}')
```

```python
# Task 2
np.dot(result_matrix, C)
```

> Множення неможливе, тому що внутрішні розмірності не збігаються (3 != 2)

```python
# Task 3
np.dot(result_matrix, D)
```

```python
# Task 1
matrix_T = result_matrix.T
print(matrix_T)
print(matrix_T.shape)

print(f'Форма - {matrix_T.shape}')
```

```python
# Task 2
vector_T = vector_uniform.T
print(vector_T)
print(f'Форма - {vector_uniform.shape}')
print(f'Форма T - {vector_T.shape}')
```

> 1D-масив не має осей “рядок/стовпець”
> - `vector` — це просто список чисел
> - `np.T` нічого не міняє, бо нема що міняти місцями

```python
# Task 3

column_vector = np.expand_dims(vector_uniform, axis=1)
print(f'Форма column_vector - {column_vector.shape}')
print(f'Форма vector_uniform - {vector_uniform.shape}')
```

> `expand_dims` додає нову вісь і тепер це вектор-стовпець

```python
# Task 4
column_vector_T = column_vector.T
print(column_vector_T)
print(f'Форма - {column_vector.shape}')
print(f'Форма T - {column_vector_T.shape}')
```
